```python
import requests
import pandas as pd

url = "https://api.open-meteo.com/v1/forecast?latitude=23.3441&longitude=85.3096&current=temperature_2m,relative_humidity_2m,wind_speed_10m"
response = requests.get(url)
data = response.json()
data
```

```python
weather = data['current']
weather
```

```python
weather = data['current']
weather['city'] = 'Ranchi'  # adding location context manually since API doesn't return it

df = pd.DataFrame([weather])  # wrapped in a list because we have one row of data
df
```

```python
# Check for null.
print(df.isnull().sum())
print(df.dtypes)
```
